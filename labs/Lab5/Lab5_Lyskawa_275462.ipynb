## Zadanie 1 - Element strukturalny. 
  
Elemnt strukturalny definuje sąsiedztwo piksela, ktróre jest brane pod uwagę podczas wyznaczania operacji morfologicznych. Bazując na definicji operacji morfologicznej oraz rozmiarze i kształcie elemntu strukturalnego możemy odpowiednio modyfikować nasze obrazy. 

### a) Ręczne ustawienie elementu strukturalnego  
 
W tym przypadku sami określimy kwadratowy element strukturalny o rozmiarze $5x5$: `kernel = np.ones((5,5),np.uint8)`.  
  
Proszę o stworzenie elemntów o rozmiarach $3x3$,$5x5$,$7x7$,$1x1$ oraz wyświetlenie ich. 

```python
import numpy as np

for size in (3, 5, 7, 1):
    kernel = np.ones((size, size), dtype=np.uint8)
    print(f"Kwadrat {size}×{size}")
    print(kernel)
```

### b) getStructuringElement 
  
Często zdaża się, że kdratowy element strukturalny nie jest odpowiedni. W celu utworzenia elementu eliptycznego można wykorzystać wbudowaną funkcję OpenCV `cv.getStructuringElement()`, gdzie należy zdefiniować kształt oraz rozmiar jądra. Więcej informacji można przeczytać w dokumentacji [tutaj](https://docs.opencv.org/4.x/d4/d86/group__imgproc__filter.html#gac2db39b56866583a95a5680313c314ad) i [tutaj](https://docs.opencv.org/4.x/d4/d86/group__imgproc__filter.html#ga7be549266bad7b2e6a04db49827f9f32) 


Proszę o stworzenie elemntów typu CROSS i ELLIPSE o rozmiarach $3x3$,$5x5$,$7x7$,$11x11$ oraz wyświetlenie ich. 

```python
import cv2 as cv

for name, shape in (("CROSS", cv.MORPH_CROSS), ("ELLIPSE", cv.MORPH_ELLIPSE)):
    for size in (3, 5, 7, 11):
        kernel = cv.getStructuringElement(shape, (size, size))
        print(f"{name} {size}×{size}")
        print(kernel)
```

## Zadanie 2 - Podstawowe operacje morfologiczne 

W zadaniach należy wykorzystać obraz utworzony na pierwszym laboratorium, bez szumu (obraz z liniami) oraz jeden z obrazów wykonanych w laboratorium). 
  
### a) Erozja 
  
Dla zadanych obrazów należy wykonać operację erozji dla 3 rodzai elementów strukturalnych oraz rozmiarów $5x5$,$7x7$,$11x11$. Proszę wykonać analizy dla obrazów binarnych oraz monochromatycznych. 

```python
from pathlib import Path

import cv2 as cv
import matplotlib.pyplot as plt

photos = Path("../../photos")
if not photos.is_dir():
    photos = Path("photos")

lines = cv.imread(str(photos / "wzorzec_paskowy.png"), cv.IMREAD_GRAYSCALE)
gray = cv.imread(str(photos / "lab5b.jpg"), cv.IMREAD_GRAYSCALE)
_, lines_bin = cv.threshold(lines, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)
_, gray_bin = cv.threshold(gray, 0, 255, cv.THRESH_BINARY + cv.THRESH_OTSU)

images = (
    ("Paski w szarości", lines),
    ("Paski binarne", lines_bin),
    ("Lab5b w szarości", gray),
    ("Lab5b binarny", gray_bin),
)
shapes = (
    ("RECT", cv.MORPH_RECT),
    ("CROSS", cv.MORPH_CROSS),
    ("ELLIPSE", cv.MORPH_ELLIPSE),
)

kernels = []
for label, shape in shapes:
    for size in (5, 7, 11):
        kernel = cv.getStructuringElement(shape, (size, size))
        kernels.append((f"{label} {size}×{size}", kernel))

fig, axes = plt.subplots(1, 4, figsize=(12, 4), layout="tight")
for ax, (name, img) in zip(axes, images):
    ax.imshow(img, cmap="gray", vmin=0, vmax=255)
    ax.set_title(name)
    ax.axis("off")
plt.show()


def show_results(name, results):
    fig, axes = plt.subplots(3, 3, figsize=(9, 8), layout="tight")
    fig.suptitle(name)
    for ax, (label, out) in zip(axes.flat, results):
        ax.imshow(out, cmap="gray", vmin=0, vmax=255)
        ax.set_title(label)
        ax.axis("off")
    plt.show()


for name, img in images:
    results = [(label, cv.erode(img, kernel)) for label, kernel in kernels]
    show_results(name, results)
```

W obu wersjach obrazu jasne paski zwęziły się, a najcieńsze zniknęły. Efekt nasilał się wraz z rozmiarem maski i wyglądał tak samo dla trzech kształtów. Na binarnym lab5b białe obszary się skurczyły, a obraz w skali szarości pociemniał. Przy 11×11 zmiana była wyraźniejsza dla RECT niz dla CROSS.

### b) Dylatacja 
  
Dla zadanych obrazów należy wykonać operację dylatacji dla 3 rodzai elementów strukturalnych oraz rozmiarów $5x5$,$7x7$,$11x11$. Proszę wykonać analizy dla obrazów binarnych oraz monochromatycznych. 

```python
for name, img in images:
    results = [(label, cv.dilate(img, kernel)) for label, kernel in kernels]
    show_results(name, results)
```

W obu wersjach obrazu jasne paski poszerzyły się. Przy większych maskach częściej się łączyły, podobnie dla wszystkich kształtów. Na binarnym lab5b białe obszary stały się szersze, a w skali szarości jasne fragmenty rozlały się na sąsiednie obszary. Przy 11×11 najlepiej widać to dla RECT.

### c) Otwarcie
  
Dla zadanych obrazów należy wykonać operację otwarcia dla 3 rodzai elementów strukturalnych oraz rozmiarów $5x5$,$7x7$,$11x11$. Proszę wykonać analizy dla obrazów binarnych oraz monochromatycznych. 

```python
for name, img in images:
    results = [(label, cv.morphologyEx(img, cv.MORPH_OPEN, kernel)) for label, kernel in kernels]
    show_results(name, results)
```

Najcieńsze jasne paski zniknęły, a pozostałe zachowały swoją szerokość. Im większa maska, tym mniej pasków zostało. Trzy kształty dały tu taki sam widoczny efekt. Na binarnym lab5b ubyło drobnych białych fragmentów, a w skali szarości osłabły niewielkie jasne szczegóły. Przy 11×11 CROSS zachował ich więcej niz RECT.

### c) Zamknięcie  
  
Dla zadanych obrazów należy wykonać operację zamknięcia dla 3 rodzai elementów strukturalnych oraz rozmiarów $5x5$,$7x7$,$11x11$. Proszę wykonać analizy dla obrazów binarnych oraz monochromatycznych. 

```python
for name, img in images:
    results = [(label, cv.morphologyEx(img, cv.MORPH_CLOSE, kernel)) for label, kernel in kernels]
    show_results(name, results)
```

W obu wersjach obrazu wąskie ciemne przerwy między paskami zostały wypełnione. Większe maski połączyły więcej pasków, podobnie dla wszystkich trzech kształtów. Na binarnym lab5b część ciemnych szczelin zniknęła, a w skali szarości drobne ciemne miejsca stały się mniej widoczne. Przy 11×11 efekt był wyraźniejszy dla RECT niz dla CROSS.
